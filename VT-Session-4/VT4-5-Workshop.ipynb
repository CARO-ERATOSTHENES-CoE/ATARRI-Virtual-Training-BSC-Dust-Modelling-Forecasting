# VT4 - Workshop

In the following workshop, you will evaluate a dust event with MONARCH dust AOD and AERONET. You may use either the Providentia dashboard or the Providentia library for your evaluation, as you prefer.

If you want to use the Providentia library, you can use this notebook to make your plots.

### Evaluation of MONARCH dust AOD against AERONET

You will evaluate MONARCH dust AOD against both AERONET SDA coarse AOD (`od500aerocoarse`) and AERONET Direct Sun total AOD (`od550aero`) filtered by the Angstrom exponent. You can select either of the following events:

1. 7-11 June 2025 Iberian dust event: Some suggested stations you can use are El_Arenosillo, Cabo_de_Roca, Evora, Madrid, etc.
2. 15-19 July 2025 trans-Atlantic dust event. Some suggested stations you can use are Dakar_Belair, Capo_Verde, Santa_Cruz_Tenerife, etc.

Observational data has been made available to you in your `/shared/trainees/atarrixx/obs/nonghost/` folder.

Once you have selected your event, carry out the following steps:

- Modify the dates in the `atarri_interpolation.conf` file and run your interpolation by using the Providentia interpolation command in the terminal. Make sure to add the `--cores=1` command when working inside the training virtual machine. 
- Launch either the dashboard or the library for both AERONET algorithms and analyse your data. Modify the dates in the `atarri_od500aerocoarse.conf` and `atarri_od550aero.conf` files to cover the period of your dust event.

For each event, you will need to analyse with both normal (mean, NData) and bias statistics (r, MFE, MFB). Use 3-4 stations to validate arrival times, magnitude of modelled and observed dust, and look at site-specific performances. 

Once you have evaluated both algorithms, compile your plots and your comments and present them to the group.
