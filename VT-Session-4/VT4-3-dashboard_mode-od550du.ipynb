# 4.3 Evaluation of dust optical depth at 550nm with Dashboard mode

Once we have access to our interpolated models and observations, we can make use of Providentia's visualisation functionalities. 

For this next exercise, we are going to evaluate dust AOD at 550nm with Dashboard mode.

## Configuration file

To evaluate our model, we will use a different configuration file `atarri_od550aero.conf`. This file specifies the full name of the interpolated model `monarch-regional-000` and also contains other parameters, most of which are normally used in report mode. We will select the interpolated data that corresponds to the `od550aero` species and specify the `directsun_v3-lev15` algorithm.

It is best practice to separate .conf files according to their intended use i.e. a separate file for interpolations and another one for evaluations in the Dashboard or Library.

![Direct Sun.png](img/12_directsun-conf.png)

### Applying the Angstrom exponent as a filter

Right now, we have selected all aerosol observations. However, mineral dust is a coarse aerosol, and in order to make a fair evaluation we need to isolate these coarser particles. This can be done by filtering the AOD column with the Angstrom exponent.

To do this, we will use `filter_species` parameter, which filters the read species (od550aero) by another species within the data range. This is set in the format `network1:species1 (>lowerlim, <=upperlim, fillvalue), network2:species2 (:, <upperlim))`. In our case, we will set this to:

~~~
filter_species = nasa-aeronet/directsun_v3-lev15:ae440-870aero (>0.6,<=200, nan)
~~~

By selecting AOD values above an Angstrom of 0.6, we will only select the coarsest particles and can reliably assume that this output will be dust.

## Launching the dashboard

To launch Dashboard mode, we can go to our Providentia directory and run `./bin/providentia` in the terminal. This will launch an empty dashboard, from where we can load any configuration file through the interface.

![Load conf in dashboard](img/13_dashboard-load.png)

We can also directly load our specific configuration file by running `./bin/providentia --config=atarri_od550aero.conf` in the terminal from the Providentia directory.

In both cases, Providentia will ask which sections specified in our .conf we want to load onto the Dashboard.

## Evaluation of dust event with dashboard

To begin, it is always a good idea to check our initial map of all stations. Two useful statistics to look for in here are the number of data points (NData) and biases (MFB, r and FGE for example).

Remember to check the burger menu and activate both Dataset 1 and Dataset 2 if you want to look for bias statistics.

![Number of data points](img/31_ndata.png)

![MFB Global](img/32_mfb-global.png)

You will notice by checking the number of data points that we do not have many in areas affected by our dust event. This is caused by the multi-species filtering: we are removing all data points that do not meet our Angstrom Exponent filtering requirement. This is important for us to note, as we may choose to not include stations in our evaluation if they have too few valid observations to ensure appropriate representativity.

Take a look at how many data points (and stations) appear if we remove the multi-species filtering.

![Data read without multi-species filtering](img/33_without-filter.png)

Now, we will select the Eastern Mediterranean as our area of interest. First, we will look at the timeseries (colocated and non-colocated) with four stations:

- Migal, Lebanon
- CUT-TEPAK, Cyprus
- AgiaMarina_Xyliatou, Cyprus
- FORTH_RSLab, Greece

The non-colocated timeseries is useful in looking at arrival times and analysing the intensity of observed vs modelled dust, whereas the colocated option contextualises how many model points we can actually match to observations. Remember, we need colocation turned on to look at bias statistics.

![Selected stations for March dust event](img/34_selected-stations.png)

![Selected stations for March dust event without colocation](img/35_selected-stations-noncolocated.png)
