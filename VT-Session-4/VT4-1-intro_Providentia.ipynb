## AERONET and in-situ observations

AERONET (full name AErosol RObotic NETwork) is a federation of ground-based remote sensing aerosol networks established by NASA and PHOTONS. This collaboration provides globally distributed observations of spectral aerosol optical depth, inversion products, and precipitable water in diverse aerosol regimes.

Version 3 AOD data are computed for three data quality levels: Level 1.0 (unscreened), Level 1.5 (cloud-screened and quality-controlled), and Level 2.0 (quality-assured). Inversions, precipitable water, and other AOD-dependent products are derived from these levels and may implement additional quality checks. You can find more details [here](https://aeronet.gsfc.nasa.gov/).

## Providentia

Providentia is a tool developed at BSC-CNS, which has been designed to generate on-the-fly analysis and reports of models, with respect to processed ground-based observational data. It is publicly accessible through the following [GitHub](https://github.com/BSC-ES/providentia) repository.

### Structure

Providentia has several functionalities, which are listed as follows:

![Providentia-structure](img/01_structure.png)

- `Download` mode: Allows the user to download both observation and model data, as defined in the configurations. This requires access to HPC accounts, and the user may download data onto their local machine.
- `Interpolation` mode: Matches the models to the observation networks of interest. It can be run on both HPC and local machines.
- `Dashboard` mode: Produces an interactive dashboard that allows the user to analyse models in real time by using observational data. It can be run both on HPC and local machines.
- `Report` mode: Creates detailed pdf reports for in-depth analysis of the model and observational data. It can be run both on HPC and local machines.
- `Library` mode: A powerful tool that allows the user to apply backend functions within scripts to reproduce other Providentia feaures. It can be run both on HPC and local machines.

As part of our training, we will focus on the `Interpolation`, `Dashboard`, and `Library` modes.

##### Directories

Providentia relies on the following directory structure to store essential files for its operation:

- `configurations/`: Contains configuration files required to run all Providentia modes.
- `logs/`: Stores output files generated by Interpolation mode.
- `notebooks/`: Holds template notebooks.
- `plots/`: Saved plots if Providentia is used as a module.
- `reports/`: Saved .pdf reports generated in `Report` mode.
- `saved_data/`: Stores configuration, NetCDF and NumPy files if Providentia is used as a module.
- `settings/`: Contains files that configure various aspects of Providentia.

#### Creating a configuration file

Configuration files are used by Providentia so it knows what data to interpolate (Interpolation mode), download (Download mode), read and plot (Dashboard, Report and Library modes).

They are read from the `configurations/` folder by default, and are composed of mandatory sections and optional subsections. When reading and filtering the data, the subsections take all the values defined in the parent sections.

Note that the sections are mandatory to launch Providentia. They must be wrapped in brackets `[ ]` and cannot include interpuncts `·`. THe following fields are mandatory to be included in a section:

- `network`: The network you want to load observations from. Multiple networks are allowed.
- `species`: The species to load. Multiple species are allowed.
- `resolution`: The temporal resolution of the observations that you want to load.
- `start_date`: The comparison start date in YYYYMMDD format.
- `end_date`: The comparison end date in YYYYMMDD format.

Optionally, if you want to set specific information, you may add a subsection to the configuration file by wrapping your inputs in double brackets `[[ ]]`. You must set them below the sections and must include all the data that you want your evaluation to focus on.
