### Comparison with remote sensing observations

Working in pairs, you will use remote sensing observations to compare a dust event. You can choose from either of the following events:

1. 15-19 July, 2025: Saharan trans-Atlantic transport, affecting Santa Cruz de Tenerife MPLNET station (latitude: 28.4720, longitude: 16.2470).
2. 7-11 June, 2025: Iberian dust intrusion, affecting El Arenosillo MPLNET station (latitude: 37.1050, longitude: 6.7340).

Your comparison will be a combination of two steps:

- Satellite comparison: Using MAPIES, process and plot your VIIRS dust AOD data. Use the frequency of your choice for the model comparison. Discuss on the spatio-temporal performance of the model compared to VIIRS (i.e. is it overestimated/underestimated around the source or during transport? What is the numerical difference? Can we make any other specific conclusions?)
- MPLNET comparison: Using the lidar processing script, create daily plots of dust extinction that compare both station and model data. Discuss the performance of the model compared to MPLNET (i.e. is it overestimated/underestimated at different times? What is the numerical difference? Can we make any other specific conclusions that we could not make with the satellite comparison?)

Make a presentation and discuss your results.
