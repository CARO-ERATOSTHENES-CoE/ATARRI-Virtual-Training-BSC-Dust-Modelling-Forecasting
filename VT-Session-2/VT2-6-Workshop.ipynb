## Suggested home exercises

Until now, we have been working with publicly available model files. In the following exercise, you will be able to work with restricted-access files. These can be found here: `/esarchive/exp/monarch/a00c_cirrus/regional/3hourly/`.

While these are also MONARCH model files, they contain the following differences:

- There are more data variables available (dust extinction (`ec550du`), dust concentration (`concdu`), wind vectors, etc.).
- 4D files have their vertical levels represented in model levels, compared to altitudes. The conversion of model level to altitude is represented with the `z` dataset (`/esarchive/exp/monarch/a00c_cirrus/regional/3hourly/z/`).
- They have a rotated grid, whereas the publicly-available files we have worked with have a regular grid. Note: if you want to subset for specific coordinates, you will need to either regrid or extract the relevant rotated latitude and longitude sets.

This means that our restricted-access files have a higher resolution than the publicly available files.

To make use of this difference in resolution, you will carry out a short qualitative analysis of modelled dust extinction. Using one of the `ec550du` datasets corresponding to the event, plot a vertical section over Limassol and compare this with your own Lidar datasets (or any other datasets of your choice). 

Optionally, you are recommended to convert the model levels to altitudes with the `z` variable.
