# 1.3 Understanding dust model outputs

#### Objective

The objective of this notebook is to provide an understanding on how to open dust model netcdf files with specific Python libraries, and what their data structure looks like.

## MONARCH .nc files

We now want to start working directly with MONARCH outputs. These are netcdf files (in .nc format) which contain information such as aerosol optical depth (AOD), aerosol dust load, and aerosol dust concentration. As part of model runs, these outputs are produced on a daily basis, and contain data at a 3-hourly frequency for a period of 72 hours, starting every day at noon. 

To facilitate their use during the coming exercises, all dust variables from the previously downloaded MONARCH files have been separated into their own .nc files. This has been done with the `/scripts/split_monarch.py` script.

To explore the data structures and metadata of these model outputs, we can use the `netCDF4` and `xarray` libraries. Although we have examples of both below, in this training series we will mostly focus on `xarray`. 

### Example 1: Dust optical depth at 550nm

First, we will look at dust AOD with `netCDF4`.

```python
import os
from pathlib import Path
from netCDF4 import Dataset
```

We will set up our root directory with `Path` and join it to our experiment output with `os`. Then, we read it with the `Dataset` class, setting the access mode to read-only.

```python
monarch_root = Path('/home/tvintimi/data/providentia/exp_to_interp/a00c_cirrus/regional/3hourly')
```

```python
monarch_dod_path = os.path.join(monarch_root, "od550_dust/od550_dust_2025030312.nc")
monarch_dod = Dataset(monarch_dod_path, mode='r')
```

If we print the variable with our stored .nc file, we can see both the general file structure and metadata. This is a 3D file, containing latitude, longitude and time variables.

```python
print(monarch_dod)
```

We can print out the specific file variables with the following line:

```python
print(monarch_dod.variables.keys())
```

We may also better understand the contents of each variable by subsetting them. For example:

```python
monarch_dod_lat = monarch_dod.variables['time'][:]
print(monarch_dod_lat)
```

### Example 2: Dust concentration

Let us now look at dust concentration with `xarray`.

```python
import xarray as xr
```

```python
monarch_conc_path = os.path.join(monarch_root, "sconc_dust/sconc_dust_2025030312.nc")
monarch_conc = xr.open_dataset(monarch_conc_path)
```

We can directly access all variables, coordinates and metadata in a table format. Dust concentration is a 4D dataset, containing the same variables as dust optical depth but also having altitude pressure levels. This also means that dust concentration files are much heavier than our other datasets.

```python
monarch_conc
```

We can also print each variable by calling it directly as a Data Array, such as the following:

```python
monarch_conc.time
```

## SDS-WAS .nc files

The SDS-WAS Multi-Model ensemble is also available as a data product from the `Data Products` of [dust.aemet.es](https://dust.aemet.es/). As we saw before, this product is a median of all the availablle dust forecasts for a particular day. To produce it, all models have been bi-linearly interpolated to a common grid mesh of 0.5 x 0.5. It is available for both AOD and concentration.

Let us take a look at the model files.

### Example 1: Dust optical depth at 550nm

```python
sdswas_root = Path('/home/tvintimi/data/providentia/exp_to_interp/multi_model/regional/3hourly')
```

```python
sdswas_dod_path = os.path.join(sdswas_root, "od550_dust/od550_dust_2025030400.nc")
sdswas_dod = xr.open_dataset(sdswas_dod_path)
sdswas_dod
```

Let us print the dust optical depth variable:

```python
sdswas_dod.OD550_DUST
```

### Example 2: Dust surface concentration

Unlike MONARCH, dust concentration from the SDS-WAS multi-model median is only available at the surface level.

```python
sdswas_conc_path = os.path.join(sdswas_root, "sconc_dust/sconc_dust_2025030400.nc")
sdswas_conc = xr.open_dataset(sdswas_conc_path)
sdswas_conc
```

```python
sdswas_conc.SCONC_DUST
```
