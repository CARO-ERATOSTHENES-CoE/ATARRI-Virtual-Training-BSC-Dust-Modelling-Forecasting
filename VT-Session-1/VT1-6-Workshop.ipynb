# VT1 - Workshop exercises

You may use the current Jupyter notebook to execute your scripts.

## 1.1 Explore the BDRC website

Using the [dust.aemet](https://dust.aemet.es/), go to the Daily Dust Products interactive map. From here, carry out the following:

- Select a time period of your choice. It must be at least 5 days.
- Select the `MONARCH` model and at least one more model of your choice.
- Look at `AOD`, `Concentration`, and `Load` and identify a dust event among these three variables.
- Select a point in your map and obtain a time series of your dust variables. Note: your time series must include the models you selected previously.
- From the same point in your map, and using `MONARCH` data, obtain a vertical profile of dust concentration.
- Check the evaluation tab for AERONET and satellite.
- Go to the Statistics tab and compare the statistics from both models.

Write a short text summarising your findings.

## 1.2 Understand the MONARCH forecast data structure

Using a different day from the provided AOD training datasets, use `xarray` to open the dataset and print out the following statistics:

- Mean
- Median
- Max
- Min

Note: you must use one day for all three provided variables (`od550_dust`, `dust_load` and `sconc_dust`). You may subset using the coordinates and altitudes of your choice.

## 1.3 Timeseries of dust variables 

Using the remaining MONARCH and the MULTI-MODEL ensemble training datasets, construct your own plot comparing the dust optical depth of different models.

Note: to subset, use the same altitude and coordinates you selected in Task 1.
