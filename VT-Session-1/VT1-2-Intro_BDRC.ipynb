#### Objective

The objective of this notebook is to provide an introduction to existing dust products available at the Barcelona Dust Regional Center, including: visualisation, evaluation, and data download tools, applied in the context of a real-life dust event.

## Why is it important to study dust?

Sand and Dust Storms (SDS) are common meteorological hazards in arid and semi-arid regions that generate large amounts of airborne mineral dust particles. Dust particles lifted by strong winds reduce visibility to near zero in regions near the sources and are regularly carried over distances of thousands of kilometres before being deposited over the land and ocean.

Dust is an essential element of the Earth’s natural biogeochemical cycles, but is also partly caused by human-induced drivers, including climate change, unsustainable land management, and water use. In turn, sand and dust storms contribute to climate change and air pollution. They have become a serious global concern in recent decades due to their significant impacts on the environment, health, agriculture and socio-economic well-being. Their impacts are felt in many regions of the world and tackling them is aligned with several targets of the Sustainable Development Goals (SDG) established by the United Nations (UN). Over the last few years, there is an increasing need for accurate information and predictions on sand and dust storms, particularly over desert regions, such as Sahara and the Middle East. The provision of access to such information is fundamental to support the development of early warning systems and mitigation plans. Additionally, it responds to the growing interest of stakeholders from different sectors, such as solar energy plant managers, aviation professionals and policymakers from the aviation, environmental and public health sectors.

Recognizing the need for international coordination of the diverse community that deals with the societal impacts of sand and dust storms, the World Meteorological Organization (WMO) initiated the development and implementation of a Sand and Dust Storm Warning Advisory and Assessment System (SDS-WAS) in 2007. The mission of the SDS-WAS is to enhance the ability of countries to deliver timely and high-quality sand and dust storm forecasts, observations and knowledge to users through an international partnership of research, operational and user communities. This collaborative international framework facilitates the transfer of technology from research to society supporting the development of the most suitable products for the relevant sectors as identified by users.

The SDS-WAS operates as an international hub of researchers, operational centres and end-users, and is organized through Regional Nodes. All the activities, initiatives and events organized within each regional node are coordinated by a Regional Steering Group (RSG) and the associated Regional Center. Currently, there are three active Regional Nodes: Northern Africa, the Middle East and Europe (NAMEE); Asia; and Pan-America and Gulf Cooperation Council.

## Barcelona Dust Regional Center (BDRC)

In view of the demand of many national meteorological services and the good results obtained by the Sand and Dust Storm Warning Advisory and Assessment System (SDS-WAS), in 2013, the first WMO "Regional Specialized Meteorological Center with activity specialization on Atmospheric Sand and Dust Forecast (RSMC-ASDF)" was created in Barcelona (Spain). This WMO Operational Center started operations in March 2014.

The objective of the WMO Barcelona Dust Operational Center is to generate and distribute operational dust predictions for Northern Africa, the Middle East and Europe. The reference model for providing operational dust forecasts to the Center is the MONARCH model, which is developed and maintained by the Barcelona Supercomputing Center (BSC-CNS).

## Case study: a dust event takes place in March 2025

Throughout this training series, we will be diving into the world of dust modeling, with a focus on a real dust event that occurred between **03/03/2025 - 08/03/2025** in the Mediterranean.

This event will serve as our case study, providing us with the opportunity to apply and refine our dust modelling skills. In each session, we will use this event to practise different aspects of dust modelling, including data analysis, visualisation, and evaluation of MONARCH dust datasets. By the end of the series, you will be equipped with the tools to work with MONARCH forecast outputs, compare forecasts with observations, and assess the accuracy of predictions in real-world scenarios.


As our starting point, we will look at the current operational dust products offered by the [BDRC website](https://dust.aemet.es/). These include:

- Forecast
- Evaluation
- Observations
- Data download

You can read more about these tools by looking at the [Tutorial Video](https://www.youtube.com/watch?v=eUFVqBMq93c) and the [User Guide](https://dust.aemet.es/products/user-guide).

### Dust Forecast Interactive Tool

We can study dust forecasts through the interactive dashboard, under the `Forecast` tab.

This dashboard has a multitude of action buttoms, which can help us customise and facilitate the analysis of our products. For example, we can change our basemaps (`VIEW` button), add layers with extra information (`LAYERS` button), change our zoom display, and download images and other information (`DOWNLOAD` button).

Importantly, we can use the panel in the left side to modify the information we are visualising. By clicking on the `Variable` tab, we open a drop down menu of different variables we can showcase. If we open the `Models` tab, we also have the option to visualise data from several forecast models.

Let's go back to our dust event. Using the panel, select the MONARCH model and one of the dust variables (AOD, Concentration, etc.). Then, use the calendar to change the dates of the forecasts (starting from 03/03/2025 until 08/03/2025). How does the dust event change through time? Which areas are affected by the dust plumes?

Tip: If you select the Concentration variable, you can also visualise forecast data at different altitudes with the bar on the right.

![Visualisation-tool](img/03_visualisation-tool.png)

In addition to map view, we can also obtain detailed model data from specific coordinates. Let's test this out by clicking over Limassol, Cyprus.

A pop-up will appear, which will show up to two options:

- `EXPLORE TIMESERIES`
- `EXPLORE VERTICAL PROFILES` (only for dust concentration).

![Explore-timeseries](img/04_explore-timeseries.png)

If we click on `EXPLORE TIMESERIES`, we obtain a line plot showcasing the concentration values over 72-hours at our specific coordinates. If we select more models, we will obtain a new line with the forecasted concentrations from each one.

![Timeseries](img/05_timeseries.png)

If we click on `EXPLORE VERTICAL PROFILES`, we obtain a vertical profile showcasing the temporal evolution of the forecast with our selected variable. Note that this feature is only available when a singular model is selected.

![Vertical-profile](img/06_vertical-profile.png)

The tab `Probability of exceedance` outputs a probability map the likelihood of exceeding a set threshold. We can change the threshold as we wish, and also select the day to which our probability will be applied.

![Exceedance](img/07_exceedance.png)

### Evaluation tool

Once we have studied our forecasts, we can look at the `Evaluation` tab.

This tab allows us to compare dust forecasts against dust-filtered observations. Specifically, they can be compared agains AERONET AOD photometric measurements and NASA/MODIS Aqua and Terra NRT AOD combined products.

In `Visual comparison`, we will set the network to AERONET and compare this with MONARCH forecasts across the timeframe we are currently studying.

![Evaluation](img/09_Evaluation.png)

As you can see, our map is then populated with AERONET network stations. Let's click on the `CUT-TEPAK` station, and explore the timeseries.

Our new window shows the comparison with AERONET and MONARCH measurements. On the right side, we have a list of all other models that can be used in the comparison. If we wanted to add another one, we simply need to click over it and a new forecast line will show in the graph. At the bottom, there is a window with which we can adjust the evaluation period.

![Evaluation-timeseries](img/10_Evaluation-timeseries.png)

If we then move to `Statistics`, we can run a similar search to perform a statistical evaluation. For our purposes, we will select the same network and model, select all available statistics, and visualise data for the entire month.

![Statistical-table](img/11_statistical-table.png)

Our output consists of summary tables with a breakdown of the model evaluation statistics, further broken down per region.

We can also use the `VIEW MAP` button to create a graphical quick-look of each statistic at each station.

![EValuation-map](img/12_evaluation-map.png)

### Observations tool

Finally, we also have the possibility to visualise key observations to track sand and dust storms, found under the `Observations` tab. Note that these datasets are different to those used for model evaluation, and are offered by partners of the BDRC.

The `EUMETSTAT RGB` product is an RGB composite that helps to monitor the evolution of dust storms during the day and night, and will only show timesteps for available images.

![EUMETSTAT-RGB](img/13_EUMETSAT-RGB.png)

The `Visibility` product shows cases of 6-hourly visibility reduction less than 5km associated with airborne sand or dust reported in METAR, SPECI, and SYNOP bulletins.

![Visibility](img/14_visibility.png)

### Data download service

WMO BDRC provides access to numerical data of available products through a THREDDS Data Server (TDS), a web server that provides metadata and data access for scientific datasets by using remote data access protocols.

In order to access and download numerical data, you will need to request user registration via [email](mailto:dust@aemet.es). You can access two types of daily dust forecasts:

- Public: provide access to numerical preductions issued more than 2 days ago.
- Restricted: provide access to all numerical predictions available for re-distribution, including those issued less than 2 days ago.

Once registration is completed, and according to the type of access that has been granted to you (Public or Restricted), you will be able to access the numerical data archive.

![Data-download](img/15_data-download.png)

The repository is divided into two folders: `BDRC THREDDS Public Data` and `BDRC THREDDS Restricted Data`.

![THREDDS](img/16_THREDDS.png)

Each folder is in turn divided into subfolders, one for each model that provides data to the Center.

![Public-data](img/17_public-data.png)

The model folders contain the published dust daily forecasts ordered per year and month. After selecting your time, you can access the list of files available for download. By selecting a specific file, the download links will appear.

![Datasets](img/18_datasets.png)

Depending on which link we access, we will be able to download an entire dataset or subset our data.

![THREDDS links](img/19_THREDDS-links.png)

We can also use the links to run download commands in our terminal. This can be done by using the following format:

~~~
wget --user="USER" --password="PASSWORD" "https://dust.aemet.es/thredds/fileServer/dataRoot/MODEL/YEAR/MONTH/FILE"

~~~

Example:

`wget --user="USER" --password="PASSWORD" "https://dust.aemet.es/thredds/fileServer/dataRoot/MONARCH/2025/03/2025031812_3H_SDSWAS_NMMB-BSC-v2_OPER.nc"`

Example with subsetting:

`wget --user="USER" --password="PASSWORD" "https://dust.aemet.es/thredds/ncss/dataRoot/MONARCH/2025/03/2025031412_3H_SDSWAS_NMMB-BSC-v2_OPER.nc?var=od550_dust&disableLLSubset=on&disableProjSubset=on&horizStride=1&time_start=2025-03-14T12%3A00%3A00Z&time_end=2025-03-18T00%3A00%3A00Z&timeStride=1&addLatLon=true&accept=netcdf"`

For future exercises, we will mostly use MONARCH datasets downloaded from the HTTPServer.
