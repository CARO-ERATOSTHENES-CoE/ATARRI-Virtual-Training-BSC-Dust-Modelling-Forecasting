# Introduction to our learning platform

## Setting up

To access your material, you will need to have internet connection. Please use the credentials provided to you by your trainers to connect to the `eduroam` network. You will be able to access your training environment regardless of your operating system.

## JupyterHub

The training material for these sessions has been pre-loaded into our common training platform: a JupyterHub.

JupyterHub is a multi-user version of Jupyter notebooks, which manages separate Jupyter environments for each user. We are running [The Littlest JupyterHub](https://tljh.jupyter.org/en/latest/), which allows us to host up to 100 users simultaneously. This JupyterHub is hosted using a virtual machine ran from BSC, which has been constructed taking into account the technical specifications needed to run our tools and support our workflow. While JupyterHub mainly focuses on supporting Jupyter notebooks, our training material will also include standard presentations, Python scripts, bash scripts, etc., all of which have already been pre-loaded onto the platform.

### Accessing the JupyterHub training environment.

You may access the training environment through the JupyterHub server link shared by your trainers.

Please use the credentials provided to you to log on.

### Accessing the virtual machine

For specific training activities, you will need to access the virtual machine via ssh connection. Your trainers will assist you with the command to access the machine.

## Environments

To support the scripts, notebooks and tools needed for our trainings, we have pre-loaded Python environments onto our JupyterHub. These are self-contained compilations of libraries and packages needed to support the execution of our scripts. For future use of these tools, you may want to install these environments in either your local or virtual machines. In the table below, we provide a list of the conda environments that we have installed and also provide either the .yaml file or link to .yaml file to install them. You can use the following command to install an environment: `conda env create -f environment.yaml`.

Notes: A Linux operating system is recommended. To create your environments, you will require either `conda`, `miniconda` or `mamba` (the last two are recommended).

| Environment | Requirements file path | VT Session |
| --- | --- | --- |
| ATARRI v0.0.1 | `conda_envs/atarri_v0.0.1/environment.yaml` | Sessions 0, 1 & 2 |
| MAPIES v0.2.0 | `conda_envs/mapies_v0.2.0/environment.yaml` | Session 3 |
| PROVIDENTIA v3.0.0 | You can find the .yaml file in the Providentia GitHub [page](https://github.com/BSC-ES/providentia/blob/master/environment.yaml) | Session 4 |

Every time you open a Jupyter notebook, you will be prompted to select a kernel from the list of available environments. Please select the appropriate kernel as instructed by your trainer. You may also use the table above as reference. In the case you need to activate an environment from the terminal of the Virtual Machine, you can run the following command: `conda activate environment`. If you experience any technical issues, please let your trainer know.

## Repository structure

Below you will find a description of the folders structure you will work with.

### /home/

Upon logging in to the platform, you will see on the left panel your home directory. Each user has their own, with read and write permissions, so that you can work simultaneously in your own environment. The path to this is `/home/jupyter-user/`. Inside this directory, each user has a folder `atarri`, inside of which you will find the following subfolders:

- `conda_envs`: Contains .yaml files to install the packages for each environment.
- `csv`: Currently empty subdirectory. It will be used to store temporary .csv files for data processing.
- `functions`: Contains Jupyter notebooks with specific Python functions required to run our code from the main training notebooks.
- `logs`: Empty folder used to store log files from MAPIES.
- `plot_outputs`: Currently empty subdirectory. It will be used to store plot outputs produced from the main training notebooks.
- `scripts`: Contains auxiliary files and scripts needed to quickly run intermediate steps in the main training notebooks.
- `VT-Session-1`: Contains Jupyter notebooks for the training session "1. Introduction to existing dust products and data structures".
- `VT-Session-2`: Contains Jupyter notebooks for the training session "2. Data visualisation".
- `VT-Session-3`: Contains Jupyter notebooks for the training session "3. Comparison of forecasts with remote sensing observations".
- `VT-Session-4`: Contains Jupyter notebooks for the training session "4. Evaluation of forecasts using AERONET and in-situ observations".
- `tools`: Contains the tools needed for sessions T3 and T4.

Inside the `tools` subfolder, you will find your own copy of the following two tools:

- `mapies`: Folder containing MAPIES, used for plotting and comparison of satellites and models.
- `providentia`: Folder containing Providentia, used for on-the-fly evaluations and analysis of model data.

### /shared/

The `/shared/` directory is available to all JupyterHub users from the terminal. This partition has the following structure:

- `/data/`: Contains all non-interpolated dust model data, satellite data and ground observations to be used from sessions T1-T4. This folder is read-only.
- `/trainees/`: Contains sub-folders for each Jupyter user to store large output datasets. Each user has exclusive read-write permissions to their own folder.
